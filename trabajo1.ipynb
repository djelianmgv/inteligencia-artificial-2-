Resumen extendido de IA y Machine Learning
1. IA: patrones complejos y no lineales
Una de las grandes capacidades de la Inteligencia Artificial (IA) y, especialmente, del Machine Learning, es encontrar patrones dentro de grandes volúmenes de información.

Cuando hablamos de patrones complejos no lineales, significa que la relación entre las variables de entrada y la variable que queremos predecir no puede representarse adecuadamente mediante una línea recta o una relación simple.

Por ejemplo:

Predecir si un cliente abandonará una empresa.
Detectar fraude en transacciones.
Reconocer imágenes.
Predecir precios.
Clasificar correos como spam o no spam.
En estos problemas puede haber muchas variables y relaciones entre ellas. Los algoritmos de ML pueden aprender esas relaciones a partir de los datos.

Idea clave:

Machine Learning busca aprender una función que transforme los datos de entrada en una predicción útil, incluso cuando la relación entre las variables es compleja y no lineal.

3. Fuga de datos — Data Leakage
Este es uno de los conceptos más importantes de tus apuntes.

La fuga de datos ocurre cuando información que no debería estar disponible durante el entrenamiento termina entrando en el proceso de aprendizaje del modelo.

Esto genera un modelo que parece tener un rendimiento excelente durante las pruebas, pero que posteriormente funciona mal con datos reales.

Ejemplo
Supongamos que queremos predecir si un paciente tendrá determinada enfermedad.

Tenemos:

Edad
Peso
Presión arterial
Resultados de pruebas
Diagnóstico final
Si utilizamos el diagnóstico final como variable durante el entrenamiento, estamos dando al modelo información que en realidad solamente conoceríamos después.

El modelo puede obtener una precisión altísima, pero sería engañosa.

¿Cómo evitar la fuga?
Una regla fundamental es:

Primero se divide el conjunto de datos y después se aprende cualquier transformación utilizando únicamente el conjunto de entrenamiento.

Por ejemplo:

Datos → Train/Test → transformar Train → aplicar transformación a Test

No:

Datos → escalar todo → Train/Test

porque el escalado de todo el conjunto podría permitir que información estadística del conjunto de prueba llegue al entrenamiento.

4. Imputación, escalamiento, compresión y clasificación
Tus apuntes:

imputación → escalamiento → comprime → clasificador

Esto representa parte del proceso de preprocesamiento y construcción de un modelo.

Imputación
La imputación consiste en tratar los valores faltantes.

Por ejemplo:

Edad	Ingreso	Compra
25	1500	Sí
31	—	No
42	3000	Sí

El valor faltante de ingreso puede sustituirse por:

Media.
Mediana.
Moda.
Un valor calculado mediante otro modelo.
Para datos numéricos, la mediana suele ser una opción robusta cuando existen valores extremos.

Escalamiento
Las variables pueden estar en escalas muy diferentes.

Ejemplo:

Edad: 18–80
Ingreso: 1.000–20.000.000
Número de compras: 0–100
Algunos algoritmos funcionan mejor cuando las variables tienen escalas comparables.

Dos métodos comunes son:

StandardScaler
Transforma los datos para que tengan aproximadamente:

Media = 0
Desviación estándar = 1
Min-Max Scaling
Lleva los valores a un intervalo, normalmente:

0 a 1

El escalamiento es especialmente importante para algoritmos basados en distancias o magnitudes, como:

KNN
K-Means
SVM
Regresión logística, dependiendo del contexto
5. Compresión / reducción de dimensionalidad
Cuando tienes cientos o miles de variables, trabajar directamente con todas puede ser costoso y algunas pueden aportar poca información.

La reducción de dimensionalidad intenta representar los datos utilizando menos variables o componentes.

Un método clásico es PCA (Principal Component Analysis).

Ejemplo:

Tenemos:

100 variables → PCA → 10 componentes

La idea es conservar la mayor cantidad posible de información relevante utilizando menos dimensiones.

Esto puede ayudar a:

Reducir el costo computacional.
Reducir ruido.
Facilitar visualizaciones.
Evitar problemas derivados de demasiadas variables.
Pero hay que distinguir compresión/reducción de dimensionalidad de simplemente eliminar variables: PCA crea nuevas variables llamadas componentes principales.

7. El "truco de kernel"
El kernel trick o truco del kernel es especialmente importante en algoritmos como Support Vector Machines (SVM).

La idea es muy interesante.

Supongamos que nuestros datos no pueden separarse mediante una línea recta.

En dos dimensiones tenemos algo así:

      ● ● ●
    ●       ●

          ○
        ○ ○ ○

Una línea recta podría no separar correctamente los grupos.

El truco del kernel permite trabajar como si los datos fueran transformados a un espacio de mayor dimensión, donde pueden ser separables de una manera más sencilla.

En lugar de calcular explícitamente esa transformación, el algoritmo utiliza una función kernel para calcular relaciones entre los puntos.

Algunos kernels conocidos:

Lineal.
Polinomial.
RBF o gaussiano.
Sigmoidal.
Idea para memorizar
Kernel = transformar implícitamente el problema para encontrar separaciones que no son evidentes en el espacio original.

8. Bagging
Tus apuntes:

BAGGING reduce usando árboles

Aquí falta una idea importante.

Bagging significa Bootstrap Aggregating.

Es una técnica de ensamble que entrena múltiples modelos sobre diferentes muestras obtenidas mediante bootstrap y después combina sus resultados.

Uno de los ejemplos más conocidos es Random Forest, que utiliza muchos árboles de decisión.

La idea:

Datos originales
      ↓
 ┌────┼────┬────┐
 ↓    ↓    ↓    ↓
Árbol Árbol Árbol Árbol
 ↓    ↓    ↓    ↓
 └────┼────┼────┘
      ↓
   combinación
      ↓
   predicción

En clasificación, los árboles pueden votar.

Ejemplo:

Árbol 1 → Clase A
Árbol 2 → Clase A
Árbol 3 → Clase B
Árbol 4 → Clase A
Árbol 5 → Clase B
Resultado:

Clase A

¿Qué consigue Bagging?
Principalmente ayuda a reducir la varianza de modelos que pueden ser muy sensibles a los datos de entrenamiento.

9. Sesgo y varianza
Este es otro concepto fundamental.

En Machine Learning queremos encontrar un equilibrio entre bias (sesgo) y variance (varianza).

Alto sesgo
El modelo es demasiado simple.

Puede ocurrir underfitting.

El modelo no aprende suficientemente los patrones.

Ejemplo:

Modelo demasiado simple
        ↓
No captura la complejidad
        ↓
Alto sesgo

Alta varianza
El modelo es demasiado sensible a los datos específicos utilizados para entrenarlo.

Puede ocurrir overfitting.

Modelo demasiado complejo
        ↓
Aprende incluso el ruido
        ↓
Alta varianza

Una forma sencilla de recordarlo:

Problema	Modelo	Resultado
Alto sesgo	Demasiado simple	Underfitting
Alta varianza	Demasiado complejo	Overfitting
Equilibrio	Complejidad adecuada	Mejor generalización.
¿Qué hace Bagging?
Una de sus principales ventajas es reducir la varianza.

Por eso Random Forest suele ser más estable que un único árbol de decisión.

10. GridSearchCV vs RandomizedSearchCV
Tu apunte dice:

GridSearchCV costosa, Randomized más rápida y no tan costosa

La idea es correcta.

Estos métodos sirven para encontrar buenos hiperparámetros de un modelo.

Los hiperparámetros son configuraciones que nosotros elegimos antes o durante el entrenamiento.

Por ejemplo, en Random Forest:

Número de árboles.
Profundidad máxima.
Número de variables consideradas por división.
GridSearchCV
Prueba todas las combinaciones que especificamos.

Supongamos:

n_estimators: 100, 200, 300 → 3 opciones
max_depth: 5, 10, 20 → 3 opciones
min_samples_split: 2, 5 → 2 opciones
Tenemos:

3 × 3 × 2 = 18 combinaciones

Si además hacemos validación cruzada de 5 particiones:

18 × 5 = 90 entrenamientos

Por eso puede ser costoso.

RandomizedSearchCV
En lugar de probar todas las combinaciones, selecciona un número determinado de combinaciones al azar dentro de las distribuciones o valores que definamos.

Por ejemplo:

10 combinaciones aleatorias × 5 folds = 50 entrenamientos

Puede ser mucho más rápido cuando tenemos un espacio de búsqueda enorme.

¿Cuál utilizar?
GridSearchCV:

Espacio de búsqueda pequeño.
Queremos revisar exhaustivamente las combinaciones.
RandomizedSearchCV:

Muchas variables/hiperparámetros.
Espacio de búsqueda grande.
Tenemos un presupuesto computacional limitado.
Queremos encontrar rápidamente buenas configuraciones.
Importante: RandomizedSearchCV no garantiza encontrar la combinación óptima, pero puede encontrar una muy buena con mucho menos costo.
